# Candidate Text (for every candidate)

```python
import json
import pandas as pd

data = []

with open(link, "r", encoding="utf-8") as f:

    for line in f:

        candidate = json.loads(line)

        # Basic Profile
        candidate_id = candidate["candidate_id"]

        title = candidate["profile"]["current_title"]

        headline = candidate["profile"]["headline"]

        summary = candidate["profile"]["summary"]

        experience = candidate["profile"]["years_of_experience"]

        # Skills
        skills = " ".join(
            [skill["name"] for skill in candidate["skills"]]
        )

        # Career History
        career_text = " ".join(
            [job["description"] for job in candidate["career_history"]]
        )

        # Final Candidate Text
        candidate_text = (
            headline
            + " "
            + summary
            + " "
            + title
            + " "
            + skills
            + " "
            + career_text
        )

        data.append({
            "candidate_id": candidate_id,
            "title": title,
            "experience": experience,
            "candidate_text": candidate_text
        })

df = pd.DataFrame(data)

print(df.head())
```

## Production Evidence score

```python
production_keywords = [
    "retrieval",
    "ranking",
    "search",
    "recommendation",
    "recommendation system",
    "vector database",
    "embedding",
    "embeddings",
    "faiss",
    "milvus",
    "pinecone",
    "weaviate",
    "qdrant",
    "opensearch",
    "elasticsearch",
    "learning to rank",
    "ab test",
    "a/b test",
    "production ml",
    "deployed"
]
```

```python
def production_score(career_text):

    text = career_text.lower()

    hits = 0

    for keyword in production_keywords:

        if keyword in text:
            hits += 1

    return hits
```

```python
behavior_df["offer_acceptance"] = (
    behavior_df["offer_acceptance"]
    .clip(lower=0)
)
```

```python
positive_phrases = [
    "built",
    "deployed",
    "production",
    "owned",
    "shipped",
    "serving",
    "queries",
    "users",
    "ndcg",
    "mrr",
    "ranking",
    "retrieval",
    "evaluation",
    "a/b",
    "ab testing",
    "monitoring",
    "mlflow",
    "kubeflow",
    "learning-to-rank"
]
```

```python
negative_phrases = [
    "online courses",
    "experimenting with",
    "side projects",
    "exploring ai",
    "generative ai explorer",
    "grow my ai capabilities",
    "excited about ai"
]
```

```python
def professional_score(text):

    text = str(text).lower()

    positive = 0
    negative = 0

    for phrase in positive_phrases:
        if phrase in text:
            positive += 1

    for phrase in negative_phrases:
        if phrase in text:
            negative += 1

    return positive - negative
```

```python
candidate_text_df["professional_score"] = (
    candidate_text_df["candidate_text"]
    .apply(professional_score)
)
```

```python
import json

profile_lookup = {}

with open(link, "r", encoding="utf-8") as f:

    for line in f:

        c = json.loads(line)

        profile_lookup[c["candidate_id"]] = {
            "title": c["profile"]["current_title"],
            "headline": c["profile"]["headline"],
            "summary": c["profile"]["summary"]
        }

print("Loaded", len(profile_lookup))
```

```python
import re

def create_reason(candidate_id):

    profile = profile_lookup[candidate_id]

    title = profile["title"]
    headline = profile["headline"]
    summary = profile["summary"]

    text = (headline + " " + summary).lower()

    reason = ""

    # ---------------------------------------
    # Years of Experience
    # ---------------------------------------
    years = ""

    match = re.search(r'(\d+(\.\d+)?)\+?\s*years', summary)

    if match:
        years = match.group(1)

    if years:
        reason += f"{title} with {years}+ years of professional experience. "
    else:
        reason += f"{title}. "

    # ---------------------------------------
    # Domain
    # ---------------------------------------

    if "nlp" in text:
        reason += "Strong background in Natural Language Processing. "

    elif "backend" in text:
        reason += "Strong backend engineering and data infrastructure experience. "

    elif "machine learning" in text:
        reason += "Strong machine learning background. "

    elif "data scientist" in text:
        reason += "Strong applied data science experience. "

    elif "computer vision" in text:
        reason += "Experience building computer vision solutions. "

    # ---------------------------------------
    # AI Technologies
    # ---------------------------------------

    tech = []

    technologies = [
        "langchain",
        "llamaindex",
        "rag",
        "semantic search",
        "embeddings",
        "faiss",
        "pinecone",
        "bm25",
        "tensorflow",
        "pytorch",
        "llms",
        "opensearch",
        "weaviate",
        "qdrant",
        "milvus",
        "recommendation",
        "retrieval",
        "ranking"
    ]

    for t in technologies:
        if t in text:
            tech.append(t.title())

    tech = tech[:4]

    if tech:
        reason += (
            "Demonstrates expertise in "
            + ", ".join(tech)
            + ". "
        )

    # ---------------------------------------
    # Production Experience
    # ---------------------------------------

    if any(word in text for word in [
        "production",
        "built",
        "deployed",
        "designed",
        "owned",
        "shipped"
    ]):

        reason += (
            "The profile highlights hands-on experience delivering production-grade AI systems. "
        )

    # ---------------------------------------
    # Evaluation
    # ---------------------------------------

    if any(word in text for word in [
        "ndcg",
        "mrr",
        "evaluation",
        "offline",
        "online"
    ]):

        reason += (
            "Demonstrates experience improving retrieval quality using evaluation metrics. "
        )

    # ---------------------------------------
    # Transition to AI
    # ---------------------------------------

    if any(word in text for word in [
        "transition",
        "transitioning",
        "learning",
        "exploring",
        "completed",
        "self-directed"
    ]):

        reason += (
            "Shows continuous learning and active transition toward modern AI technologies. "
        )

    # ---------------------------------------
    # Final Recommendation
    # ---------------------------------------

    reason += (
        "Overall, the candidate demonstrates strong technical capability and relevant experience, making the profile well aligned with the target role."
    )

    return reason
```
